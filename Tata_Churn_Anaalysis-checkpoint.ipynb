##  Core Analytical Findings & Insights

```python
Churn_counts = df['Churn'].value_counts()
print("persentage of custumer doing Churn: ", round((Churn_counts['Yes']/ Churn_counts.sum())* 100, 2))
```

```python
churn_grouping = df.groupby('Churn')
```

```python
avg_tenure = churn_grouping['tenure'].median()
```

```python
print(avg_tenure)
```

```python
crossover = pd.crosstab(df['Contract'], df['Churn'])
print(crossover)
```

```python
avg_monthlycharges = churn_grouping['MonthlyCharges'].mean()
print(avg_monthlycharges)
```

```python
phone_crossover = pd.crosstab(df['PhoneService'], df['Churn'])
print(phone_crossover)
```

```python
internet_crossover = pd.crosstab(df['InternetService'], df['Churn'] )
security_crossover = pd.crosstab(df['OnlineSecurity'],df['Churn'])
```

```python
multiple_crossover = pd.crosstab(index = [df['InternetService'], df['OnlineSecurity']], columns = df['Churn'])
print("===Internet Cross Table===")
print(internet_crossover)
print("percentage of Churn(DSL): ", round(internet_crossover.loc['DSL', 'Yes']/(internet_crossover.loc['DSL', 'Yes']+internet_crossover.loc['DSL', 'No']) * 100, 2), "%")
print("percentage of Churn(fiber optic): ", round(internet_crossover.loc['Fiber optic', 'Yes']/(internet_crossover.loc['Fiber optic', 'Yes']+internet_crossover.loc['Fiber optic', 'No']) * 100, 2), "%")
print("percentage of Churn(No): ", round(internet_crossover.loc['No', 'Yes']/(internet_crossover.loc['No', 'Yes']+internet_crossover.loc['No', 'No']) * 100, 2), "%")

print("\n\n===Cecurity Cross Table===")
print(security_crossover)
print("percentage of Churn(No): ", round(security_crossover.loc['No', 'Yes']/(security_crossover.loc['No', 'Yes']+security_crossover.loc['No', 'No']) * 100, 2), "%")
print("percentage of Churn(No internet service): ", round(security_crossover.loc['No internet service', 'Yes']/(security_crossover.loc['No internet service', 'Yes']+security_crossover.loc['No internet service', 'No']) * 100, 2), "%")
print("percentage of Churn(Yes): ", round(security_crossover.loc['Yes', 'Yes']/(security_crossover.loc['Yes', 'Yes']+security_crossover.loc['Yes', 'No']) * 100, 2), "%")

print("\n\n===Both Togather Cross Table===")
print(multiple_crossover)
```
